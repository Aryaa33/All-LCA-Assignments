```python
#list
fruits=["Apple", "Banana","Papaya","Guava"]
print("\nfruits:",fruits)
```

```python
#tuple
fruits=("Apple", "Banana","Papaya","Guava")
print("\nfruits:",fruits)
```

```python
#dictionary
fruits={
    101:"Apple", 
    102:"Banana",
    103:"Papaya",
    104:"Guava"
}
print("\nfruits:",fruits)
```

```python
#dictionary after updating
fruits[102]="Strawberry"
print("\nfruits after updating :",fruits)
```

```python
#dictionary after adding 105
fruits[105]="Raspberry"
print("\nfruits after adding:",fruits)
```

```python
#dictionary after deleting 103
del fruits[103]
print("\nfruits after deleting:",fruits)
```
