```python
#basic calculator
num1=float(input("Enter the first no:"))
num2=float(input("Enter the second no:"))
print("\nchoose an operator:,1.add,2.substract,3.divide,4.multiply")
choice=int(input("Enter your choice(1-4):"))
if choice==1:
    print("\nthe addition of the numbers provided is:",num1+num2)
elif choice==2:
    print("\nthe substraction of the numbers provide is:",num1-num2)
elif choice==3:
    print("\nthe division of the numbers provided is:",num1/num2)
else:
    print("\nthe multiplication of the numbers provide is:",num1*num2)
```
